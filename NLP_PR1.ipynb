### 1.1 What is NLP and Why We Need It
Natural Language Processing (NLP) is a subfield of artificial intelligence and computer science that enables computers to read, interpret, analyse, and generate human language. Computers natively process structured numerical data (tables, sensor readings, pixel arrays), whereas human communication occurs almost entirely in unstructured text and speech filled with ambiguity, idioms, and context-dependent grammar. Every day, billions of messages, support tickets, product reviews, and social media posts are generated at a scale that no human team could manually read or categorise. NLP bridges this gap by transforming messy, unstructured text into structured mathematical representations that algorithms can process automatically in milliseconds.

**Five Real-World Applications of NLP (Including Indian Use Cases):**
1. **Customer Review & Social Media Sentiment Analysis (Indian Use Case — Swiggy, Zomato & IndiGo):** Indian food-delivery platforms like Swiggy and Zomato and airlines like IndiGo process millions of customer feedback messages in English and Hinglish daily to detect delivery delays, food quality issues, or flight disruption complaints automatically.
2. **Conversational AI & Multilingual Voice Assistants (Indian Use Case — IRCTC AskDisha & Bhashini / UPI Voice):** Indian Railways' AskDisha chatbot and the Government of India's Digital India [Bhashini](https://bhashini.gov.in/) platform allow citizens to book tickets and access services in regional languages such as Hindi, Gujarati, Tamil, and Marathi.
3. **Spam & Phishing Detection:** Email providers (Gmail, Outlook) and telecom filters scan incoming messages for deceptive patterns, suspicious links, and urgency keywords to block spam and financial fraud before it reaches the inbox.
4. **Search Engines & Information Retrieval:** Google Search and enterprise search engines parse user queries, match synonyms, and rank billions of web pages based on semantic relevance rather than exact keyword matching alone.
5. **Machine Translation:** Tools like Google Translate convert text across hundreds of language pairs in real time, preserving grammatical structure and context across cultures.


### 1.2 Common NLP Tasks and Approaches

#### Table 1: Common NLP Tasks
| NLP Task | One-Line Description | Real-World Example |
| :--- | :--- | :--- |
| **Text Classification** | Assigns a predefined category label to an entire document or message. | Routing an airline support email to *Baggage Claim*, *Refunds*, or *Flight Booking*. |

| **Machine Translation** | Automatically converts text from a source human language into a target human language. | Translating a passenger safety notice from English into Hindi or Gujarati. |
| **Text Summarisation** | Condenses a long document into a short summary while preserving key facts. | Summarising 500 passenger survey responses into a three-bullet executive brief. |
| **Question Answering (QA)** | Extracts or generates a direct answer to a user's natural-language question from a knowledge base. | Answering *"What is the baggage allowance for domestic economy flights?"* from airline policy docs. |

#### Class Imbalance & Metric Selection Discussion
- **Strong Class Imbalance:** The dataset is heavily skewed toward complaints: roughly **62.7% negative**, **21.1% neutral**, and only **16.2% positive** tweets.
- **Why Accuracy Alone is Misleading:** A trivial majority-class classifier that predicts `"negative"` for every tweet would achieve ~63% accuracy while failing completely (`0%` recall) on neutral and positive tweets.
- **Why Macro-F1 is Our Primary Metric:** **Macro-F1** computes the F1-score (harmonic mean of Precision and Recall) independently for each of the three classes and then takes their unweighted average. Because `positive` (16%) and `negative` (63%) contribute equally to Macro-F1, a model can only achieve a high Macro-F1 if it performs well across all three sentiment classes.

# Task 2: Text Cleaning with RegEx & Tokenisation
**Objective:** Design and test a sequential Regular Expression (RegEx) cleaning pipeline, preserve negation words during stop-word removal, and inspect how preprocessing changes the vocabulary distribution.  
**Expected Output:** Tested RegEx patterns with an explanatory Markdown table, a `clean_text` column with zero unhandled empty rows, before/after tokenisation examples preserving negations, and class-wise vocabulary frequency plots.


### 2.1 Designing the Text Cleaning Rules

1. **Lowercase Conversion (`s.lower()`):** Unifies case variations (`"Delayed"`, `"DELAYED"`, `"delayed"`) into a single token `"delayed"` so the vocabulary does not fragment across casing styles.
2. **Remove URLs (`https?://\S+|www\.\S+`):** Hyperlinks point to external pages or shortened `t.co` tracking codes that behave as unique noise tokens with zero generalisable sentiment value.
3. **Remove `@mentions` (`@\w+`):** Every tweet begins with an airline handle (`@united`, `@VirginAmerica`, `@AmericanAir`). If kept, the classifier would memorise airline-specific base rates (a form of shortcut learning) instead of learning linguistic sentiment cues.
4. **Keep Hashtag Words, Drop `#` Symbol (`#(\w+) -> \1`):** Hashtags in tweets often carry the core sentiment word (e.g., `#neveragain`, `#worstflight`, `#greatservice`). Keeping the word while dropping `#` merges `#delayed` with `delayed`.
5. **Decode HTML Entities & Expand Negation Contractions (`won't -> will not`, `n't -> not`):** Before stripping punctuation, expanding `didn't` to `did not` ensures that the negation word `not` survives intact rather than turning into `didnt`.
6. **Remove Numbers and Punctuation (`[^a-z\s]`):** Strips flight numbers (`1245`), gate codes, and punctuation marks so only clean alphabetic words remain.
7. **Collapse Extra Whitespace (`\s+`):** Replaces multiple consecutive spaces left behind by removed tokens with a single space and strips leading/trailing whitespace.

> **What Useful Information Might Be Lost?**  
> Stripping punctuation removes repeated exclamation marks (`"!!!"`), question marks (`"???"`), and emoticons (`":("`, `":)"`), and lowercasing erases `ALL CAPS` shouting (e.g., `"WORST AIRLINE EVER"`). Similarly, removing non-ASCII characters removes emojis (`✈️`, `😡`, `❤️`), which often convey strong emotion in short tweets.


````python
# 2.2 RegEx Practice Cells — Test Each Pattern Individually Before Combining
sample_raw = "@VirginAmerica Flight #402 is DELAYED again!!! Check https://t.co/xyz123 #WorstService # Delayed :(   "

print("Original Sample :", repr(sample_raw))
print("1. Lowercase    :", repr(sample_raw.lower()))
print("2. Remove URLs  :", repr(re.sub(r"https?://\S+|www\.\S+", " ", sample_raw.lower())))
print("3. No @mentions :", repr(re.sub(r"@\w+", " ", sample_raw.lower())))
print("4. Hashtag word :", repr(re.sub(r"#(\w+)", r"\1", sample_raw.lower())))
print("5. No numbers   :", repr(re.sub(r"\d+", " ", sample_raw.lower())))
print("6. Alpha only   :", repr(re.sub(r"[^a-z\s]", " ", sample_raw.lower())))
print("7. Collapse ws  :", repr(re.sub(r"\s+", " ", sample_raw.lower()).strip()))
````

#### RegEx Pattern Reference Table
| RegEx Pattern | Replacement | Meaning of Each Symbol | Example Match (`Before` → `After`) |
| :--- | :--- | :--- | :--- |
| `https?://\S+` | `" "` | `http` literal, `s?` optional `'s'`, `://` literal, `\S+` one or more non-whitespace chars | `"See https://t.co/ab1"` → `"See  "` |
| `@\w+` | `" "` | `@` symbol followed by `\w+` (one or more word characters `[a-zA-Z0-9_]`) | `"@AmericanAir help"` → `"  help"` |
| `#(\w+)` | `\1` | `#` symbol followed by capture group `(\w+)`; replaced by captured word `\1` | `"#CustomerService"` → `"CustomerService"` |
| `n't\b` | `" not"` | Contraction suffix `n't` at a word boundary `\b` expanded before stripping apostrophes | `" wasn't happy"` → `" was not happy"` |
| `[^a-z\s]` | `" "` | Negated character class `[^...]`: matches anything that is NOT a lowercase letter `a-z` or whitespace `\s` | `"flight #402!!!"` → `"flight     "` |
| `\s+` | `" "` | Matches 1 or more whitespace characters (spaces, tabs, newlines) | `"delayed    again"` → `"delayed again"` |

#### Why Blindly Removing All Stop-Words Can Reverse Sentiment
Standard stop-word lists (including NLTK's English list) are designed for topic modelling and information retrieval, so they include function words like `"not"`, `"no"`, `"nor"`, and `"never"`. In sentiment analysis, however, negation words act as **polarity modifiers**: removing `"not"` from `"not happy with the delay"` leaves `["happy", "delay"]`, making a complaint look partially positive! By explicitly subtracting negation words (`no`, `not`, `nor`, `never`, etc.) from our stop-word set, we keep the crucial negation signal intact for both heuristic rules and N-gram feature extractors.

#### Vocabulary Observations & Implications for `min_df` and `max_features`
1. **Surprising Shared Words:** Domain nouns like `"flight"`, `"get"`, `"cancelled"`, and `"service"` appear with very high frequency across **both** negative and neutral/positive classes. A pure unigram model that overweights `"flight"` will struggle unless common domain words are down-weighted via **TF-IDF**.
2. **Long Tail of Singletons (Hapax Legomena):** Roughly half of the cleaned vocabulary consists of words that appear **only once** in the entire dataset (misspellings, rare proper nouns, confirmation codes).
3. **Choosing `min_df` and `max_features` in Task 4:** Setting `min_df=2` (or `min_df=3`) immediately prunes thousands of one-off spelling typos that cannot generalise to unseen test tweets, while capping `max_features=5000` retains all high-signal unigrams and bigrams while keeping memory and overfitting low.

# Task 4: Feature Extraction — Bag-of-Words, TF-IDF & N-Grams
**Objective:** Understand how raw text is converted into numerical feature matrices by implementing Bag-of-Words and TF-IDF from scratch, verifying them against `scikit-learn`, and evaluating vocabulary vs. memory trade-offs across N-gram ranges.  
**Expected Output:** Comparison table of feature extraction techniques, manual vs. `sklearn` verification for BoW and TF-IDF (`np.allclose == True`), training/test matrix sparsity statistics, top class bigrams, and a feature-size vs. memory chart.


### 4.1 Why Feature Extraction Is Needed & Technique Comparison
Machine learning classifiers (Logistic Regression, Naive Bayes, SVMs) operate on numerical vectors $\mathbf{x} \in \mathbb{R}^d$ and cannot multiply raw character strings by weights. Feature extraction maps each variable-length text document into a fixed-length numeric vector across a shared vocabulary derived from the training corpus.

| Feature Technique | What It Captures | What It Ignores | Vector Size ($d$) | Sparsity | Typical Weakness |
| :--- | :--- | :--- | :--- | :--- | :--- |
| **Bag-of-Words (BoW)** | Exact raw count of each vocabulary word in a document. | Word order, grammar, context, and semantic similarity between synonyms. | Vocabulary size $|V|$ (typically $5,000 - 15,000$). | Extremely high ($>99\%$ zeros). | Frequent domain words (`"flight"`, `"airline"`) dominate counts even though they carry little sentiment signal. |
| **TF-IDF** | Word frequency in the document scaled by how rare/informative the word is across the whole corpus. | Word order and synonym relationships (`"delayed"` vs `"late"` remain orthogonal columns). | Vocabulary size $|V|$ ($5,000 - 15,000$). | Extremely high ($>99\%$ zeros). | Still treats every word as an independent axis and fails on word-order inversions (`"good, not bad"` vs `"bad, not good"`). |
| **N-Grams ($N=2, 3$)** | Contiguous sequences of $N$ words (e.g., `"not happy"`, `"customer service"`, `"never again"`). | Long-range dependencies separated by more than $N$ words and semantic similarity across different phrases. | Explodes rapidly ($50,000 - 200,000+$ features). | Even higher ($>99.9\%$ zeros). | Massive vocabulary growth increases memory consumption and risk of overfitting on rare phrase combinations. |


````python
# 4.2 Bag-of-Words by Hand and with scikit-learn CountVectorizer
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

tiny_corpus = [
    "the flight was good not bad",
    "the flight was bad not good",
    "great flight and great crew",
    "not happy with the delayed flight"
]

# 1. Manual Bag-of-Words using Python Counter
vocab_sorted = sorted({word for doc in tiny_corpus for word in doc.split()})
manual_bow_rows = []
for doc in tiny_corpus:
    counts = Counter(doc.split())
    manual_bow_rows.append([counts.get(w, 0) for w in vocab_sorted])

df_bow_manual = pd.DataFrame(manual_bow_rows, columns=vocab_sorted, index=[f"Doc {i+1}" for i in range(4)])

# 2. scikit-learn CountVectorizer
cv_tiny = CountVectorizer()
sk_bow_matrix = cv_tiny.fit_transform(tiny_corpus).toarray()
df_bow_sklearn = pd.DataFrame(sk_bow_matrix, columns=cv_tiny.get_feature_names_out(), index=[f"Doc {i+1}" for i in range(4)])

print("Manual Bag-of-Words DataFrame:")
display(df_bow_manual)
print("scikit-learn CountVectorizer DataFrame:")
display(df_bow_sklearn)
print("Exact match between Manual BoW and CountVectorizer:", df_bow_manual.equals(df_bow_sklearn))
````

#### Why `"the movie was good, not bad"` and `"the movie was bad, not good"` Are Identical in Unigram BoW
As seen in `Doc 1` and `Doc 2` above, both sentences contain the exact same multiset of words (`{"bad": 1, "flight": 1, "good": 1, "not": 1, "the": 1, "was": 1}`). Because unigram Bag-of-Words discards word order completely, both sentences produce the **exact same feature vector** `[0, 1, 0, 0, 1, 0, 1, 0, 1, 1, 1, 0]`, making it mathematically impossible for any classifier using unigram BoW to tell them apart!


````python
# 4.3 TF-IDF by Hand vs TfidfVectorizer(smooth_idf=False, norm=None) vs Default TfidfVectorizer()
N_docs = len(tiny_corpus)
tf_matrix = df_bow_manual.values.astype(float)

# Document frequency df(t): number of documents containing term t
df_counts = (tf_matrix > 0).sum(axis=0)

# Textbook/unsmoothed formula in sklearn: idf(t) = ln(N / df(t)) + 1
idf_manual = np.log(N_docs / df_counts) + 1.0
tfidf_manual = tf_matrix * idf_manual

# Compare with TfidfVectorizer(smooth_idf=False, norm=None)
tfidf_unsmoothed = TfidfVectorizer(smooth_idf=False, norm=None)
sk_tfidf_unsmoothed = tfidf_unsmoothed.fit_transform(tiny_corpus).toarray()

print("np.allclose(Manual TF-IDF, TfidfVectorizer(smooth_idf=False, norm=None)):",
      np.allclose(tfidf_manual, sk_tfidf_unsmoothed))

df_tfidf_manual = pd.DataFrame(np.round(tfidf_manual, 4), columns=vocab_sorted, index=[f"Doc {i+1}" for i in range(4)])
print("\nManual TF-IDF Matrix (smooth_idf=False, norm=None):")
display(df_tfidf_manual)

# Compare with default TfidfVectorizer() (smooth_idf=True, norm='l2')
tfidf_default = TfidfVectorizer()
sk_tfidf_default = tfidf_default.fit_transform(tiny_corpus).toarray()
df_tfidf_default = pd.DataFrame(np.round(sk_tfidf_default, 4), columns=tfidf_default.get_feature_names_out(), index=[f"Doc {i+1}" for i in range(4)])
print("\nDefault TfidfVectorizer() Output (smooth_idf=True, norm='l2'):")
display(df_tfidf_default)
````

#### Explaining the Two Differences in Default `TfidfVectorizer()` and Low Weight of Ubiquitous Words
1. **IDF Smoothing (`smooth_idf=True`):** Default `scikit-learn` adds `1` to both the numerator and denominator inside the logarithm: $\text{idf}(t) = \ln\left(\frac{1 + N}{1 + \text{df}(t)}\right) + 1$, acting as if an extra document containing every word in the vocabulary exists so zero-division can never occur.
2. **L2 Normalisation (`norm='l2'`):** Each document row vector $\mathbf{v}$ is divided by its Euclidean norm $\|\mathbf{v}\|_2 = \sqrt{\sum_i v_i^2}$ so every document vector has unit length ($\|\mathbf{v}\|_2 = 1$). This prevents long tweets from having artificially larger feature values than short tweets.
3. **Why Words Appearing in Every Document Get a Low Weight:** In our 4 sentences, `"flight"` appears in all 4 documents ($\text{df} = N = 4$), so $\ln(4/4) + 1 = 0 + 1 = 1.0$ (the minimum possible IDF). Because `"flight"` appears everywhere, knowing that a tweet contains `"flight"` gives zero discriminative signal about whether the tweet is positive or negative.


````python
# 4.4 BoW and TF-IDF on the Airline Tweets (Fitted on TRAINING Data ONLY)
def matrix_stats(name: str, X_tr, X_te, vocab_len: int) -> dict:
    total_cells = X_tr.shape[0] * X_tr.shape[1]
    nnz = X_tr.nnz
    sparsity = 100.0 * (1.0 - nnz / total_cells)
    return {
        "Vectorizer Setting": name,
        "Train Shape": str(X_tr.shape),
        "Test Shape": str(X_te.shape),
        "Vocabulary Size": vocab_len,
        "Train Non-Zero (nnz)": f"{nnz:,}",
        "Sparsity (%)": round(sparsity, 4),
        "Train Data Memory (KB)": round(X_tr.data.nbytes / 1024, 2)
    }

# Full vocabulary vs min_df=2, max_features=5000
cv_full = CountVectorizer()
X_tr_bow_full = cv_full.fit_transform(X_train_clean)
X_te_bow_full = cv_full.transform(X_test_clean)

tv_full = TfidfVectorizer()
X_tr_tfidf_full = tv_full.fit_transform(X_train_clean)
X_te_tfidf_full = tv_full.transform(X_test_clean)

cv_pruned = CountVectorizer(min_df=2, max_features=5000)
X_tr_bow_pruned = cv_pruned.fit_transform(X_train_clean)
X_te_bow_pruned = cv_pruned.transform(X_test_clean)

tv_pruned = TfidfVectorizer(min_df=2, max_features=5000)
X_tr_tfidf_pruned = tv_pruned.fit_transform(X_train_clean)
X_te_tfidf_pruned = tv_pruned.transform(X_test_clean)

stats_df = pd.DataFrame([
    matrix_stats("CountVectorizer (Default Full)", X_tr_bow_full, X_te_bow_full, len(cv_full.vocabulary_)),
    matrix_stats("TfidfVectorizer (Default Full)", X_tr_tfidf_full, X_te_tfidf_full, len(tv_full.vocabulary_)),
    matrix_stats("CountVectorizer (min_df=2, max_features=5000)", X_tr_bow_pruned, X_te_bow_pruned, len(cv_pruned.vocabulary_)),
    matrix_stats("TfidfVectorizer (min_df=2, max_features=5000)", X_tr_tfidf_pruned, X_te_tfidf_pruned, len(tv_pruned.vocabulary_)),
])
display(stats_df)
````

#### Why the Vectoriser MUST Be Fitted on Training Data Only
Fitting a vectoriser on the full dataset before splitting introduces **data leakage**:
1. Words that appear only in the test set would enter the vocabulary dictionary (`vocabulary_`).
2. For `TfidfVectorizer`, the inverse document frequency ($\text{idf}$) weights would be computed using document counts from the test set, leaking test-set distributional statistics into the training features and producing overly optimistic evaluation metrics that fail in real-world production.


````python
# 4.5 N-Grams: Compare (1,1), (1,2), and (1,3) & Top 15 Bigrams per Class
ngram_configs = [(1, 1), (1, 2), (1, 3)]
ngram_summary = []

for n_range in ngram_configs:
    vec = TfidfVectorizer(ngram_range=n_range)
    X_tr_ng = vec.fit_transform(X_train_clean)
    ngram_summary.append({
        "ngram_range": str(n_range),
        "Vocabulary Size": f"{len(vec.vocabulary_):,}",
        "Non-Zero Entries": f"{X_tr_ng.nnz:,}",
        "Memory (KB)": round(X_tr_ng.data.nbytes / 1024, 2)
    })

print("Vocabulary Growth Across N-Gram Ranges:")
display(pd.DataFrame(ngram_summary))

# Extract Top 15 Highest-Weighted Bigrams ONLY (ngram_range=(2,2)) for Negative and Positive Classes
vec_bigrams_only = TfidfVectorizer(ngram_range=(2, 2), min_df=3)
X_tr_bi = vec_bigrams_only.fit_transform(X_train_clean)
bi_features = np.array(vec_bigrams_only.get_feature_names_out())

neg_mask = (y_train.values == "negative")
pos_mask = (y_train.values == "positive")

neg_mean_tfidf = np.asarray(X_tr_bi[neg_mask].mean(axis=0)).ravel()
pos_mean_tfidf = np.asarray(X_tr_bi[pos_mask].mean(axis=0)).ravel()

top15_neg_idx = neg_mean_tfidf.argsort()[::-1][:15]
top15_pos_idx = pos_mean_tfidf.argsort()[::-1][:15]

top_bigrams_df = pd.DataFrame({
    "Rank": range(1, 16),
    "Top Negative Bigrams": [f"{bi_features[i]} ({neg_mean_tfidf[i]:.4f})" for i in top15_neg_idx],
    "Top Positive Bigrams": [f"{bi_features[i]} ({pos_mean_tfidf[i]:.4f})" for i in top15_pos_idx],
})
display(top_bigrams_df)
````

#### Value of Bigrams vs. Cost of Larger $N$
- **Two Concrete Bigram Examples Carrying Meaning Separate Words Do Not:**
  1. **`"customer service"` / `"cancelled flight"` / `"on hold"`:** Individually, `"customer"`, `"service"`, and `"hold"` are neutral nouns/verbs, but together `"customer service"` and `"on hold"` in airline tweets strongly signal a passenger complaint.
  2. **`"not happy"` / `"thank you"` / `"great flight"`:** In `"not happy"`, the unigram `"happy"` is positive, whereas the bigram `"not happy"` captures the exact negated sentiment polarity.

The feature matrix becomes $>99.98\%$ sparse, consumes 3–5× more RAM, slows down training, and increases overfitting on rare 3-word sequences that occur only once in the training set.


````python
# 4.6 Feature-Size Experiment: Vocabulary Size & Matrix Memory vs ngram_range and max_features
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Experiment A: Varying ngram_range (with min_df=2)
ng_labels = ["(1,1)", "(1,2)", "(1,3)"]
ng_vocabs, ng_mem_kb = [], []
for nr in [(1, 1), (1, 2), (1, 3)]:
    v = TfidfVectorizer(ngram_range=nr, min_df=2)
    X_tmp = v.fit_transform(X_train_clean)
    ng_vocabs.append(len(v.vocabulary_))
    ng_mem_kb.append(X_tmp.data.nbytes / 1024.0)

ax1 = axes[0]
ax1_mem = ax1.twinx()
l1 = ax1.plot(ng_labels, ng_vocabs, marker="o", color="#2980b9", linewidth=2.5, label="Vocabulary Size")
l2 = ax1_mem.plot(ng_labels, ng_mem_kb, marker="s", color="#e74c3c", linestyle="--", linewidth=2.5, label="Matrix Memory (KB)")
ax1.set_title("(a) Effect of ngram_range (min_df=2)", fontsize=12, fontweight="bold")
ax1.set_xlabel("ngram_range")
ax1.set_ylabel("Vocabulary Size", color="#2980b9")
ax1_mem.set_ylabel("Training Matrix Memory (KB)", color="#e74c3c")
lines_a = l1 + l2
ax1.legend(lines_a, [l.get_label() for l in lines_a], loc="upper left")

# Experiment B: Varying max_features on ngram_range=(1,2)
mf_values = [1000, 3000, 5000, 10000]
mf_vocabs, mf_mem_kb = [], []
for mf in mf_values:
    v = TfidfVectorizer(ngram_range=(1, 2), min_df=2, max_features=mf)
    X_tmp = v.fit_transform(X_train_clean)
    mf_vocabs.append(len(v.vocabulary_))
    mf_mem_kb.append(X_tmp.data.nbytes / 1024.0)

ax2 = axes[1]
ax2_mem = ax2.twinx()
l3 = ax2.plot([str(m) for m in mf_values], mf_vocabs, marker="o", color="#27ae60", linewidth=2.5, label="Vocabulary Size")
l4 = ax2_mem.plot([str(m) for m in mf_values], mf_mem_kb, marker="s", color="#8e44ad", linestyle="--", linewidth=2.5, label="Matrix Memory (KB)")
ax2.set_title("(b) Effect of max_features (ngram_range=(1,2))", fontsize=12, fontweight="bold")
ax2.set_xlabel("max_features")
ax2.set_ylabel("Effective Vocabulary Size", color="#27ae60")
ax2_mem.set_ylabel("Training Matrix Memory (KB)", color="#8e44ad")
lines_b = l3 + l4
ax2.legend(lines_b, [l.get_label() for l in lines_b], loc="upper left")

plt.tight_layout()
plt.savefig("screenshots/task4_feature_size_experiment.png", dpi=150, bbox_inches="tight")
plt.show()
````

#### Choosing a Sensible Feature-Extraction Setting
Using `ngram_range=(1,2)` with `min_df=2` (and optionally `max_features=5000` to `10000`) provides the best balance: it captures critical two-word sentiment phrases (`"not happy"`, `"cancelled flight"`, `"thank you"`) while pruning one-off spelling noise (`min_df=2`) and avoiding the steep memory jump of trigrams `(1,3)`.

### 5.1 Word Embeddings — Dense Meaning Representations
In Bag-of-Words and TF-IDF, every word is represented as a separate, orthogonal one-hot dimension: the cosine similarity between `"delayed"` and `"late"` is **zero**, just like the similarity between `"delayed"` and `"banana"`. Word embeddings solve this by mapping each word to a short, **dense vector of real numbers** (for example, $d = 100$ dimensions) learned by a shallow neural network. Guided by the distributional hypothesis — *"a word is known by the company it keeps"* (J.R. Firth, 1957) — words that appear in similar surrounding contexts (such as `"delayed"`, `"late"`, and `"cancelled"`, or `"great"`, `"awesome"`, and `"amazing"`) converge to nearby points in the 100-dimensional vector space.


### 5.2 CBoW vs. Skip-gram Architectures

| Feature / Property | Continuous Bag-of-Words (CBoW, `sg=0`) | Skip-gram (`sg=1`) |
| :--- | :--- | :--- |
| **Core Learning Task** | Predicts the **single centre target word** $w_t$ from the average of its surrounding context words. | Predicts **each surrounding context word** $w_{t+j}$ given only the single centre word $w_t$. |
| **Input → Output** | Context words $(w_{t-k}, \dots, w_{t+k}) \rightarrow$ Centre word $w_t$. | Centre word $w_t \rightarrow$ Context words $(w_{t-k}, \dots, w_{t+k})$. |
| **Training Speed** | **Faster** (averages context vectors into a single prediction step per window). | **Slower** (performs multiple separate context-word updates per window). |
| **Performance on Frequent Words** | **Better** and faster on very frequent words (`"flight"`, `"thanks"`). | Slightly noisier on ultra-frequent words unless subsampling is applied. |
| **Performance on Rare Words** | **Weaker** (averaging context smooths over rare word signals). | **Much Better** (every rare word gets direct gradient updates from each of its context pairs). |
| **When to Prefer** | Large corpora or when training speed is the primary constraint. | Smaller corpora (like our 11k training tweets) or when domain-specific/rare words matter most. |

#### Worked Window = 2 Example: `"the flight was delayed again"`
Tokens: `["the", "flight", "was", "delayed", "again"]` with context window size $C = 2$ on each side:

| Centre Word ($w_t$) | CBoW Training Pair `(Context Words → Centre Word)` | Skip-gram Training Pairs `(Centre Word → Each Context Word)` |
| :--- | :--- | :--- |
| **`the`** | `(["flight", "was"] → "the")` | `("the" → "flight")`, `("the" → "was")` |
| **`flight`** | `(["the", "was", "delayed"] → "flight")` | `("flight" → "the")`, `("flight" → "was")`, `("flight" → "delayed")` |
| **`was`** | `(["the", "flight", "delayed", "again"] → "was")` | `("was" → "the")`, `("was" → "flight")`, `("was" → "delayed")`, `("was" → "again")` |
| **`delayed`** | `(["flight", "was", "again"] → "delayed")` | `("delayed" → "flight")`, `("delayed" → "was")`, `("delayed" → "again")` |
| **`again`** | `(["was", "delayed"] → "again")` | `("again" → "was")`, `("again" → "delayed")` |


````python
# 5.3 Train Both CBoW (sg=0) and Skip-gram (sg=1) Models on TRAINING Tokens Only
from gensim.models import Word2Vec

train_tokens_list = X_train_tokens.tolist()
test_tokens_list = X_test_tokens.tolist()

# Train CBoW (sg=0)
t0 = time.time()
w2v_cbow = Word2Vec(
    sentences=train_tokens_list,
    vector_size=100,
    window=5,
    min_count=3,
    sg=0,
    epochs=20,
    seed=42,
    workers=1
)
cbow_time = time.time() - t0

# Train Skip-gram (sg=1)
t0 = time.time()
w2v_sg = Word2Vec(
    sentences=train_tokens_list,
    vector_size=100,
    window=5,
    min_count=3,
    sg=1,
    epochs=20,
    seed=42,
    workers=1
)
sg_time = time.time() - t0

w2v_summary = pd.DataFrame([
    {
        "Model": "Word2Vec CBoW (sg=0)",
        "Vocabulary Size": len(w2v_cbow.wv),
        "Vector Matrix Shape": str(w2v_cbow.wv.vectors.shape),
        "Training Time (s)": round(cbow_time, 3)
    },
    {
        "Model": "Word2Vec Skip-gram (sg=1)",
        "Vocabulary Size": len(w2v_sg.wv),
        "Vector Matrix Shape": str(w2v_sg.wv.vectors.shape),
        "Training Time (s)": round(sg_time, 3)
    }
])
display(w2v_summary)
````

````python
# 5.4 Explore the Learned Vectors: Top 5 Most Similar Words for 6 Query Words + 3 Cosine Similarities
query_words = ["delayed", "flight", "thanks", "rude", "luggage", "great"]
sim_rows = []

for qw in query_words:
    cbow_sims = [f"{w} ({s:.2f})" for w, s in w2v_cbow.wv.most_similar(qw, topn=5)] if qw in w2v_cbow.wv else ["OOV"]
    sg_sims = [f"{w} ({s:.2f})" for w, s in w2v_sg.wv.most_similar(qw, topn=5)] if qw in w2v_sg.wv else ["OOV"]
    sim_rows.append({
        "Query Word": qw,
        "CBoW Top 5 Most Similar (sg=0)": ", ".join(cbow_sims),
        "Skip-gram Top 5 Most Similar (sg=1)": ", ".join(sg_sims)
    })

print("Top 5 Nearest Neighbours in CBoW vs Skip-gram:")
display(pd.DataFrame(sim_rows))

# Compute cosine similarity of 3 word pairs
word_pairs = [("delayed", "late"), ("good", "great"), ("flight", "luggage")]
pair_rows = []
for w1, w2 in word_pairs:
    c_sim = w2v_cbow.wv.similarity(w1, w2) if (w1 in w2v_cbow.wv and w2 in w2v_cbow.wv) else np.nan
    s_sim = w2v_sg.wv.similarity(w1, w2) if (w1 in w2v_sg.wv and w2 in w2v_sg.wv) else np.nan
    pair_rows.append({
        "Word Pair": f"'{w1}' vs '{w2}'",
        "CBoW Cosine Similarity": round(float(c_sim), 4),
        "Skip-gram Cosine Similarity": round(float(s_sim), 4)
    })

print("\nCosine Similarity of 3 Word Pairs:")
display(pd.DataFrame(pair_rows))
````

#### Are the Neighbours Sensible? (Small-Corpus Limitation)
Yes, the learned neighbours capture clear domain semantics: `"delayed"` clusters with `"late"`, `"cancelled"`, and `"hrs"`, `"thanks"` clusters with `"thank"` and `"appreciate"`, and `"luggage"` clusters with `"bag"`, `"bags"`, and `"baggage"`. However, we must state honestly that our training set contains only **~11,500 short tweets (~110,000 tokens)**, whereas industrial Word2Vec models are trained on billions of tokens (e.g., Google News 100B words). Because our corpus is small, rarer words have noisier co-occurrence statistics. Between the two models, **Skip-gram (`sg=1`)** produces cleaner, more semantically coherent neighbours on this small corpus because it updates every context pair individually.

````python
# 5.6 From Word Vectors to Document Vectors (Using Skip-gram Based on 5.4 Evaluation)
def doc_vector(tokens: list, model: Word2Vec) -> np.ndarray:
    """Average the word vectors of tokens present in model.wv; return zeros if none present."""
    valid_vecs = [model.wv[w] for w in tokens if w in model.wv]
    if not valid_vecs:
        return np.zeros(model.vector_size, dtype=np.float32)
    return np.mean(valid_vecs, axis=0)

X_train_w2v = np.vstack([doc_vector(toks, w2v_sg) for toks in train_tokens_list])
X_test_w2v = np.vstack([doc_vector(toks, w2v_sg) for toks in test_tokens_list])

print(f"X_train_w2v shape (Skip-gram averaged vectors): {X_train_w2v.shape}")
print(f"X_test_w2v shape  (Skip-gram averaged vectors): {X_test_w2v.shape}")
````

#### What Information Is Lost When Averaging Word Vectors?
1. **Word Order & Syntax:** Averaging is commutative ($\frac{\mathbf{v}_a + \mathbf{v}_b}{2} = \frac{\mathbf{v}_b + \mathbf{v}_a}{2}$), so compositional structure and negation scope (which word `"not"` modifies) are lost.
2. **Dilution of Strong Sentiment Words in Long Tweets:** When a 20-word tweet contains 18 neutral operational words (`"flight"`, `"from"`, `"boston"`, `"to"`, `"dallas"`) and only 1 or 2 sharp sentiment words (`"rude"`, `"lost"`), averaging pulls the document vector toward the centroid of neutral words, washing out the extreme sentiment signal.

### 6.1 What is Text Classification & Why Use a `scikit-learn` Pipeline?
**Text Classification** is the supervised learning task of mapping an input document $d$ to one class label $y \in \{\text{negative}, \text{neutral}, \text{positive}\}$. Our end-to-end classification pipeline flows through six stages:
1. **Raw Text Input:** Noisy user tweet (`"@united delayed 3 hrs and lost my bag!"`).
2. **RegEx Cleaning (`clean_text`):** Normalises case, removes URLs/`@mentions`, expands contractions, strips punctuation.
3. **Feature Extraction (`Vectorizer`):** Converts cleaned text into a sparse TF-IDF unigram+bigram vector.
4. **Classifier (`LogisticRegression`):** Computes class log-odds $\mathbf{W}\mathbf{x} + \mathbf{b}$ and softmax probabilities.
5. **Prediction:** Selects the class with the highest probability.
6. **Evaluation:** Measures Macro-F1, Accuracy, Precision, Recall, and Confusion Matrix on held-out test data.

**Why Wrapping Vectoriser + Classifier in `sklearn.pipeline.Pipeline` Matters:**
- **Prevents Data Leakage in Cross-Validation:** Inside `cross_val_score`, a `Pipeline` fits the `TfidfVectorizer` **only on the 4 training folds** and transforms the validation fold in every split automatically.
- **Single-File Deployment:** Saving the `Pipeline` with `joblib.dump()` bundles both the fitted vocabulary/IDF weights and the classifier coefficients into one file (`sentiment_model.joblib`), guaranteeing that `app.py` applies the exact same feature mapping at inference time.


````python
# 6.2 Build and Evaluate the Baseline Pipeline: TF-IDF (1,2) + LogisticRegression(class_weight='balanced')
import sklearn
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.svm import LinearSVC

baseline_pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=2)),
    ("clf", LogisticRegression(max_iter=1000, class_weight="balanced", random_state=RANDOM_STATE))
])

t0 = time.time()
baseline_pipeline.fit(X_train_clean, y_train)
baseline_train_time = time.time() - t0

y_pred_base = baseline_pipeline.predict(X_test_clean)
base_acc = accuracy_score(y_test, y_pred_base)
base_macro_f1 = f1_score(y_test, y_pred_base, average="macro")

print(f"Baseline Pipeline Test Accuracy : {base_acc:.4f}")
print(f"Baseline Pipeline Test Macro-F1 : {base_macro_f1:.4f}")
print(f"Training Time                   : {baseline_train_time:.3f} s\n")
print("Classification Report (Baseline TF-IDF (1,2) + Balanced Logistic Regression):")
print(classification_report(y_test, y_pred_base, digits=4))

plt.figure(figsize=(6, 4.8))
cm_base = confusion_matrix(y_test, y_pred_base, labels=["negative", "neutral", "positive"])
sns.heatmap(cm_base, annot=True, fmt="d", cmap="Greens",
            xticklabels=["negative", "neutral", "positive"],
            yticklabels=["negative", "neutral", "positive"])
plt.title("Confusion Matrix — Baseline TF-IDF (1,2) + LogReg", fontsize=12, fontweight="bold")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.tight_layout()
plt.savefig("screenshots/task6_baseline_confusion_matrix.png", dpi=150, bbox_inches="tight")
plt.show()
````

#### Why `class_weight='balanced'` Is Essential for This Dataset
Because negative tweets outnumber positive tweets by roughly **4 to 1** (63% vs 16%) and neutral tweets by **3 to 1** (63% vs 21%), an unweighted loss function penalises mistakes on the minority classes much less than mistakes on the majority negative class. Setting `class_weight='balanced'` weights each class inversely proportional to its frequency ($w_c = \frac{N}{K \cdot N_c}$), forcing Logistic Regression to pay equal attention to `neutral` and `positive` tweets and substantially boosting minority-class recall and **Macro-F1**.


````python
# 6.3 Compare 6 Feature + Classifier Combinations on the Same Train/Test Split
models_to_compare = [
    (
        "BoW (Unigrams, min_df=2)",
        "Multinomial Naive Bayes",
        Pipeline([("vec", CountVectorizer(min_df=2)), ("clf", MultinomialNB())]),
        "text"
    ),
    (
        "BoW (Unigrams, min_df=2)",
        "Logistic Regression (balanced)",
        Pipeline([("vec", CountVectorizer(min_df=2)), ("clf", LogisticRegression(max_iter=1000, class_weight="balanced", random_state=42))]),
        "text"
    ),
    (
        "TF-IDF Unigrams (1,1)",
        "Logistic Regression (balanced)",
        Pipeline([("vec", TfidfVectorizer(ngram_range=(1, 1), min_df=2)), ("clf", LogisticRegression(max_iter=1000, class_weight="balanced", random_state=42))]),
        "text"
    ),
    (
        "TF-IDF Unigrams+Bigrams (1,2)",
        "Logistic Regression (balanced)",
        Pipeline([("vec", TfidfVectorizer(ngram_range=(1, 2), min_df=2)), ("clf", LogisticRegression(max_iter=1000, class_weight="balanced", random_state=42))]),
        "text"
    ),
    (
        "TF-IDF Unigrams+Bigrams (1,2)",
        "Linear SVM (balanced)",
        Pipeline([("vec", TfidfVectorizer(ngram_range=(1, 2), min_df=2)), ("clf", LinearSVC(class_weight="balanced", random_state=42))]),
        "text"
    ),
    (
        "Word2Vec Skip-gram (Averaged 100d)",
        "Logistic Regression (balanced)",
        LogisticRegression(max_iter=1000, class_weight="balanced", random_state=42),
        "w2v"
    ),
]

comparison_rows = []
trained_pipelines = {}

for feat_name, clf_name, model_obj, input_type in models_to_compare:
    t0 = time.time()
    if input_type == "text":
        model_obj.fit(X_train_clean, y_train)
        t_train = time.time() - t0
        preds = model_obj.predict(X_test_clean)
    else:
        model_obj.fit(X_train_w2v, y_train)
        t_train = time.time() - t0
        preds = model_obj.predict(X_test_w2v)

    acc = accuracy_score(y_test, preds)
    mf1 = f1_score(y_test, preds, average="macro")
    prec, rec, _, _ = precision_recall_fscore_support(
        y_test, preds, labels=["negative", "neutral", "positive"], zero_division=0
    )
    key = f"{feat_name} + {clf_name}"
    trained_pipelines[key] = (model_obj, preds)
    comparison_rows.append({
        "Features": feat_name,
        "Classifier": clf_name,
        "Accuracy": round(acc, 4),
        "Macro-F1": round(mf1, 4),
        "Precision (negative)": round(prec[0], 4),
        "Recall (positive)": round(rec[2], 4),
        "Training time (s)": round(t_train, 3)
    })

ml_comparison_df = pd.DataFrame(comparison_rows)
best_idx = ml_comparison_df["Macro-F1"].idxmax()
print(f"Best ML Combination by Macro-F1: {ml_comparison_df.loc[best_idx, 'Features']} + {ml_comparison_df.loc[best_idx, 'Classifier']} "
      f"(Macro-F1 = {ml_comparison_df.loc[best_idx, 'Macro-F1']:.4f})")
display(ml_comparison_df)
````

````python
# 6.4 Master Comparison Across All Three NLP Approach Families (Heuristic, API-Based, ML Models)
prec_h, rec_h, _, _ = precision_recall_fscore_support(y_test, y_pred_heur, labels=["negative", "neutral", "positive"], zero_division=0)
prec_v, rec_v, _, _ = precision_recall_fscore_support(y_test, y_pred_vader, labels=["negative", "neutral", "positive"], zero_division=0)

heur_row = {
    "Features": "Heuristic Lexicon (74 domain words + negation)",
    "Classifier": "Hand-crafted Rules",
    "Accuracy": round(accuracy_score(y_test, y_pred_heur), 4),
    "Macro-F1": round(f1_score(y_test, y_pred_heur, average="macro"), 4),
    "Precision (negative)": round(prec_h[0], 4),
    "Recall (positive)": round(rec_h[2], 4),
    "Training time (s)": 0.0
}
vader_row = {
    "Features": "API-Based Ready-Made Lexicon",
    "Classifier": "NLTK VADER",
    "Accuracy": round(accuracy_score(y_test, y_pred_vader), 4),
    "Macro-F1": round(f1_score(y_test, y_pred_vader, average="macro"), 4),
    "Precision (negative)": round(prec_v[0], 4),
    "Recall (positive)": round(rec_v[2], 4),
    "Training time (s)": 0.0
}

master_df = pd.concat([pd.DataFrame([heur_row, vader_row]), ml_comparison_df], ignore_index=True)
display(master_df)

# Bar Chart of Macro-F1 Across All Approaches
plt.figure(figsize=(12, 5.5))
short_labels = [
    "Heuristic\n(Lexicon)",
    "API-Based\n(VADER)",
    "BoW +\nNaive Bayes",
    "BoW +\nLogReg",
    "TF-IDF (1,1)\n+ LogReg",
    "TF-IDF (1,2)\n+ LogReg",
    "TF-IDF (1,2)\n+ LinearSVC",
    "Word2Vec (SG)\n+ LogReg"
]
bar_colors = ["#95a5a6", "#f39c12", "#3498db", "#2980b9", "#1abc9c", "#27ae60", "#16a085", "#8e44ad"]
bars = plt.bar(short_labels, master_df["Macro-F1"], color=bar_colors, edgecolor="black")
for bar, val in zip(bars, master_df["Macro-F1"]):
    plt.text(bar.get_x() + bar.get_width() / 2, val + 0.01, f"{val:.3f}", ha="center", va="bottom", fontweight="bold")

plt.ylim(0, max(master_df["Macro-F1"]) * 1.15)
plt.ylabel("Test Macro-F1 Score", fontsize=12)
plt.title("Macro-F1 Comparison Across All NLP Approaches (Test Set)", fontsize=14, fontweight="bold")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.savefig("screenshots/task6_macro_f1_comparison.png", dpi=150, bbox_inches="tight")
plt.show()
````

#### Error Patterns & Two Concrete Improvement Ideas (Within NLP Fundamentals Scope)
**Observed Error Patterns:**
1. **Neutral Confused with Negative (Most Frequent):** Factual questions about flight changes, gate numbers, or baggage policies contain operational words (`"cancelled"`, `"delayed"`, `"baggage"`, `"change"`) that heavily correlate with negative complaints.
2. **Sarcasm & Polite Complaints:** Tweets that say `"thanks"` or `"great"` while complaining about a delay cause competing positive and negative feature weights.
3. **Very Short Tweets (1–3 Cleaned Tokens):** After stripping `@mentions` and URLs, very short tweets lack enough context words for confident classification.
4. **Mixed Sentiment Tweets:** Passengers who praise the flight attendants (`"great crew"`) while complaining about a late arrival (`"2 hour delay"`) express two sentiments in one tweet.

**Two Concrete Improvements Within This Practical's Scope:**
1. **Preserve Punctuation & Case Sentiment Features Before Stripping (`?`, `!`, ALL-CAPS count):** Because factual questions almost always end with `?` (`neutral`) while angry complaints use `!` or `ALL CAPS` (`negative`), replacing `?` with a special token `qmark_token` and `!` with `excl_token` inside `clean_text()` helps separate neutral inquiries from negative complaints.
2. **Tune Sublinear TF Scaling (`sublinear_tf=True`) and Trigrams (`ngram_range=(1,3)`, `min_df=3`):** Using logarithmic term frequency $1 + \ln(\text{tf})$ and common trigrams like `"thanks for nothing"` or `"what time is"` improves accuracy on sarcastic and question-style tweets.

### 7.1 What Is Part-of-Speech (POS) Tagging & Why It Is Useful
**Part-of-Speech (POS) Tagging** is the process of assigning a grammatical category label (such as noun, verb, adjective, adverb) to every token in a sentence based on both its lexical definition and its surrounding syntactic context.

| Grammatical Category | Penn Treebank Tag Code(s) | Function in a Sentence | Airline Tweet Example |
| :--- | :--- | :--- | :--- |
| **Noun** | `NN` (singular), `NNS` (plural), `NNP` (proper) | Names a person, place, object, or concept. | `"flight"` (`NN`), `"bags"` (`NNS`), `"Dallas"` (`NNP`) |
| **Verb** | `VB` (base), `VBD` (past), `VBG` (gerund), `VBN` (past part.) | Describes an action, occurrence, or state of being. | `"cancel"` (`VB`), `"waited"` (`VBD`), `"waiting"` (`VBG`) |
| **Adjective** | `JJ` (base), `JJR` (comparative), `JJS` (superlative) | Modifies or describes a noun (often carries strong sentiment). | `"rude"` (`JJ`), `"worse"` (`JJR`), `"worst"` (`JJS`), `"great"` (`JJ`) |
| **Adverb** | `RB` (base), `RBR` (comparative), `RBS` (superlative) | Modifies a verb or adjective (intensifiers and negations). | `"never"` (`RB`), `"not"` (`RB`), `"extremely"` (`RB`) |
| **Pronoun** | `PRP` (personal), `PRP$` (possessive) | Substitutes for a noun phrase. | `"you"` (`PRP`), `"they"` (`PRP`), `"my"` (`PRP$`) |
| **Preposition / Conjunction** | `IN` | Links nouns/clauses in time, space, or logic. | `"on"` (`IN`), `"at"` (`IN`), `"because"` (`IN`) |

**Why POS Tagging Is Useful:**
1. **Word-Sense Disambiguation:** Distinguishes whether `"delay"` or `"book"` is used as an action verb (`VB`) or an object noun (`NN`).
2. **Information & Opinion Extraction:** Helps extract `Adjective + Noun` pairs (e.g., `"rude staff"`, `"comfortable seat"`, `"broken screen"`) to pinpoint *what* customers are praising or complaining about.
3. **Feature Selection:** Allows filtering out purely structural words while retaining sentiment-heavy parts of speech (adjectives, adverbs, and verbs).


### 7.2 How POS Tagging Works & Contextual Disambiguation
1. **Rule-Based Taggers:** Start with a dictionary of allowed tags per word and apply hundreds of hand-written morphological and contextual rules (e.g., *"if a word ends in `-ing` and follows a form of `to be`, tag it as `VBG`"*).
2. **Statistical / Machine-Learning Taggers (HMM & Averaged Perceptron):** Learn from a human-annotated corpus (such as the Penn Treebank Wall Street Journal corpus) two sets of probabilities/weights:
   - **Emission / Lexical Evidence:** How likely word $w_i$ has tag $t_i$ (e.g., `"delayed"` is usually `VBN` or `JJ`).
   - **Transition / Contextual Evidence:** How likely tag $t_i$ follows the preceding tags $(t_{i-2}, t_{i-1})$ (e.g., after a determiner `DT` like `"the"` and before a noun `NN` like `"flight"`, the word in between is almost certainly an adjective `JJ`, whereas after `"was"` it is a verb `VBN`).
3. **Which Tagger NLTK's `nltk.pos_tag` Uses:** `nltk.pos_tag` uses a pre-trained **Greedy Averaged Perceptron** tagger (`averaged_perceptron_tagger`) trained on the **Penn Treebank (Wall Street Journal news articles)**. Because it was trained on formal, well-punctuated financial news, it can make mistakes on informal, lowercased tweets that lack standard capitalization and punctuation.


````python
# 7.3 Tag 5 Cleaned Tweets (Full Sentences Without Stop-Word Removal) + 2 Ambiguity Pairs
# Note: We use clean_text.split() (full sentence tokens without stop-word removal) so the tagger has grammatical context
sample_5_tweets = df["clean_text"].iloc[:5].tolist()

ambiguity_sentences = [
    "please book my flight early tomorrow",
    "i read a great book on the plane",
    "they will park the plane at the gate",
    "we walked through the central park today"
]

tagged_rows = []
for i, s in enumerate(sample_5_tweets):
    toks = s.split()
    pos_pairs = nltk.pos_tag(toks)
    tagged_rows.append({
        "Source": f"Tweet {i+1}",
        "Sentence": s,
        "POS Tagged (word/TAG)": "  ".join(f"{w}/{t}" for w, t in pos_pairs)
    })

for i, s in enumerate(ambiguity_sentences):
    toks = s.split()
    pos_pairs = nltk.pos_tag(toks)
    tagged_rows.append({
        "Source": f"Ambiguity Example {i+1}",
        "Sentence": s,
        "POS Tagged (word/TAG)": "  ".join(f"{w}/{t}" for w, t in pos_pairs)
    })

display(pd.DataFrame(tagged_rows))
````

#### Analysis of Context Disambiguation & Tagging Errors on Noisy Tweets
- **Context Disambiguation Works:** In `"please book my flight"` vs. `"a great book on the plane"`, the Averaged Perceptron tagger uses the preceding determiner+adjective (`"a great"`) vs. imperative context to distinguish `"book"` as a verb (`VB`/`NN`) vs. a singular noun (`NN`). Similarly, in `"they will park/VB the plane"` (following modal auxiliary `"will"/MD`) vs. `"the central park/NN today"` (following determiner+adjective `"the central"`), context resolves `"park"` into `VB` vs. `NN`.
- **Two Examples of Tagging Errors on Cleaned Tweets:**
  1. **Lowercased Proper Nouns Mislabelled as Common Nouns or Adjectives (`NN`/`JJ` instead of `NNP`):** Because `clean_text()` lowercases the text, airline or city names (like `"virgin"`, `"dallas"`, `"united"`) lose their capital first letter, causing the WSJ-trained tagger (which relies heavily on capitalization for `NNP`) to tag them as `JJ` or `NN`.
  2. **Slang & Stripped Apostrophes (`"cant"`, `"im"`, `"ive"`):** When apostrophes are removed from `"can't"` or `"i'm"`, tokens like `"im"` or `"ive"` are unfamiliar to the formal news corpus and often get mislabelled as nouns (`NN`) or adjectives (`JJ`) instead of pronoun+verb (`PRP` + `VBP`).


````python
# 7.4 POS Distribution by Sentiment on a Sample of 2,000 Training Tweets
train_sample_df = pd.DataFrame({
    "clean_text": X_train_clean,
    "sentiment": y_train
}).sample(n=min(2000, len(X_train_clean)), random_state=RANDOM_STATE)

pos_counts_by_class = {s: Counter() for s in ["negative", "neutral", "positive"]}
adj_by_class = {s: Counter() for s in ["negative", "neutral", "positive"]}
total_tokens_by_class = Counter()

for _, row in train_sample_df.iterrows():
    sent = row["sentiment"]
    toks = str(row["clean_text"]).split()
    tagged = nltk.pos_tag(toks)
    total_tokens_by_class[sent] += max(len(tagged), 1)
    for word, tag in tagged:
        if tag.startswith("JJ"):
            pos_counts_by_class[sent]["Adjectives (JJ*)"] += 1
            if word not in {"not", "no", "i", "u"} and len(word) > 2:
                adj_by_class[sent][word] += 1
        elif tag.startswith("RB"):
            pos_counts_by_class[sent]["Adverbs (RB*)"] += 1
        elif tag.startswith("VB"):
            pos_counts_by_class[sent]["Verbs (VB*)"] += 1
        elif tag.startswith("NN"):
            pos_counts_by_class[sent]["Nouns (NN*)"] += 1

pos_prop_rows = []
for sent in ["negative", "neutral", "positive"]:
    tot = total_tokens_by_class[sent]
    pos_prop_rows.append({
        "Sentiment": sent,
        "Adjectives (JJ*) %": round(100.0 * pos_counts_by_class[sent]["Adjectives (JJ*)"] / tot, 2),
        "Adverbs (RB*) %": round(100.0 * pos_counts_by_class[sent]["Adverbs (RB*)"] / tot, 2),
        "Verbs (VB*) %": round(100.0 * pos_counts_by_class[sent]["Verbs (VB*)"] / tot, 2),
        "Nouns (NN*) %": round(100.0 * pos_counts_by_class[sent]["Nouns (NN*)"] / tot, 2),
    })

pos_dist_df = pd.DataFrame(pos_prop_rows).set_index("Sentiment")
display(pos_dist_df)

# Plot Grouped Bar Chart of POS Proportions by Sentiment
ax = pos_dist_df.plot(kind="bar", figsize=(10, 5), edgecolor="black", colormap="viridis")
ax.set_title("POS Tag Proportions Across Sentiment Classes (2,000 Training Tweets)", fontsize=13, fontweight="bold")
ax.set_ylabel("Percentage of Total Tokens (%)")
ax.set_xlabel("Sentiment Class")
ax.tick_params(axis="x", rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.savefig("screenshots/task7_pos_distribution.png", dpi=150, bbox_inches="tight")
plt.show()

# Print 10 Most Frequent Adjectives for Negative and Positive Classes
top10_adj_df = pd.DataFrame({
    "Rank": range(1, 11),
    "Top 10 Negative Adjectives (JJ*)": [f"{w} ({c})" for w, c in adj_by_class["negative"].most_common(10)],
    "Top 10 Positive Adjectives (JJ*)": [f"{w} ({c})" for w, c in adj_by_class["positive"].most_common(10)],
})
display(top10_adj_df)
````

#### Observations on POS Distribution Across Sentiment Classes
1. **Positive Tweets Have the Highest Adjective (`JJ*`) Share:** Passengers writing positive tweets use descriptive evaluative adjectives (`"great"`, `"good"`, `"best"`, `"awesome"`, `"friendly"`, `"helpful"`, `"amazing"`) much more frequently than in neutral or negative tweets.
2. **Negative Tweets Have Higher Verb (`VB*`) and Adverb (`RB*`) Proportions:** Negative complaints describe actions, failures, and negations (`"delayed"`, `"cancelled"`, `"waiting"`, `"stuck"`, `"lost"`, `"not"`, `"never"`, `"still"`).
3. **Neutral Tweets Are Noun-Heavy (`NN*`):** Neutral inquiries focus on factual entities (flight numbers, cities, times, dates, tickets, gates) with fewer emotional adjectives or intensifiers.


````python
# 7.5 POS-Filtered Features (Keep Only Adjectives JJ*, Adverbs RB*, Verbs VB*, and Negation Words)
ALLOWED_POS_PREFIXES = ("JJ", "RB", "VB")

def pos_filter_text(cleaned_str: str) -> str:
    toks = str(cleaned_str).split()
    if not toks:
        return "empty_tweet"
    tagged = nltk.pos_tag(toks)
    kept = [
        w for w, tag in tagged
        if tag.startswith(ALLOWED_POS_PREFIXES) or w in NEGATION_WORDS
    ]
    return " ".join(kept) if kept else "empty_tweet"

print("Filtering training and test sets by POS tags (JJ*, RB*, VB* + negations)...")
t0 = time.time()
X_train_pos_filtered = X_train_clean.apply(pos_filter_text)
X_test_pos_filtered = X_test_clean.apply(pos_filter_text)
pos_filter_time = time.time() - t0

pos_pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=2)),
    ("clf", LogisticRegression(max_iter=1000, class_weight="balanced", random_state=RANDOM_STATE))
])
pos_pipeline.fit(X_train_pos_filtered, y_train)
y_pred_pos = pos_pipeline.predict(X_test_pos_filtered)

pos_compare_df = pd.DataFrame([
    {
        "Pipeline Variant": "Full Cleaned Text + TF-IDF (1,2) + LogReg (Task 6.2)",
        "Accuracy": round(base_acc, 4),
        "Macro-F1": round(base_macro_f1, 4),
        "Vocabulary Size": len(baseline_pipeline.named_steps["tfidf"].vocabulary_)
    },
    {
        "Pipeline Variant": "POS-Filtered (JJ*, RB*, VB* + Negations) + TF-IDF (1,2) + LogReg",
        "Accuracy": round(accuracy_score(y_test, y_pred_pos), 4),
        "Macro-F1": round(f1_score(y_test, y_pred_pos, average="macro"), 4),
        "Vocabulary Size": len(pos_pipeline.named_steps["tfidf"].vocabulary_)
    }
])
display(pos_compare_df)
````

#### Honest Evaluation: Did POS Filtering Help, Hurt, or Make No Difference?
POS filtering **slightly lowers Macro-F1 and Accuracy** compared to the full-text pipeline (while also adding several seconds of POS-tagging overhead at inference time). There are two clear reasons why:
1. **Domain Nouns Carry Strong Sentiment in Airline Tweets:** Filtering out nouns (`NN*`) removes high-signal domain terms like `"thanks"`, `"delay"`, `"delays"`, `"luggage"`, `"baggage"`, `"hours"`, `"hold"`, and `"service"`. In airline complaints, nouns often state the problem directly!
2. **Broken Bigram Context:** Removing nouns breaks informative bigrams like `"customer service"`, `"cancelled flight"`, and `"two hours"`, reducing the discriminative power of `(1,2)` N-gram features.
